```python
import chess
import chess.svg
import chess.pgn
import ipywidgets as widgets
from IPython.display import display
```

```python
def my_analysis(pgn: chess.pgn.Game):
    board = pgn.board()
    mainline_moves = pgn.mainline_moves()
    engine = chess.engine.SimpleEngine.popen_uci(r"E:\\chess\\stockfish_15_win_x64_popcnt\\stockfish_15_x64_popcnt.exe")
    scores = list()
    for i in mainline_moves:
        info = engine.analyse(board, chess.engine.Limit(time=0.1), multipv=1)
        scores.append(info[0]["score"])
        board.push(i)

    engine.quit()
    return(scores)
```

```python
def player_centipawn_loss(scores, player: str = "ZannahRain"):
    if pgn.headers["White"] == player:
        a2 = [i.white().score(mate_score=100) for i in a]  # Scores from white's perspective
        first_player_move = 0
    elif pgn.headers["Black"] == player:
        a2 = [i.black().score(mate_score=100) for i in a]  # Scores from white's perspective
        first_player_move = 1

    a3 = [max(min(i, 500), -500) for i in a2]  # Bounds
    a4 = [a3[i] - a3[i-1] for i in range(1, len(a3))]  # Difference in score from one move to the next
    a5 = [max(min(i, 500), -500) for i in a4]  # Bounds
    a6 = [-i for i in a5]  # centipawn LOSS, not gain

    white_centipawn_loss = a6[first_player_move::2]
    return(white_centipawn_loss)

all_centipawn_losses = list()
for i in range(50):
    with open(f"my_pgns/{i}.pgn", "r") as f:
        pgn = chess.pgn.read_game(f)
    print("...")
    a = my_analysis(pgn)
    all_centipawn_losses.extend(player_centipawn_loss(a))
```

```python
import plotly.express as px
import pandas as pd
df = pd.DataFrame({"all_centipawn_losses": all_centipawn_losses})
fig = px.histogram(all_centipawn_losses, nbins=50)
fig.show()
```

```python
import pandas as pd
df = pd.DataFrame({"all_centipawn_losses": all_centipawn_losses})
df["all_centipawn_losses"] = df["all_centipawn_losses"].apply(lambda x: max(x, 0))
df["group"] = pd.cut(df["all_centipawn_losses"], bins=3)
df["weight"] = (df["all_centipawn_losses"] / 100) ** 2
df2 = df.groupby("group")[["weight"]].sum()
df2["loss"] = df.groupby("group")["all_centipawn_losses"].mean()
df2 = df2.reset_index()
fig1 = px.bar(df2, x="loss", y="weight")
fig1.show()
```
